```python
for i in range(1, 5):
    if i % 2 == 0:
        print(i, '짝수입니다')
    else:
        print(i, '홀수입니다')
```

함수

```python
def square(x):
    p2 = x * x
    return p2

x1 = 12
r1 = square(x1)
print(x1, r1)
```

```python
def square(x):
    p2 = x * x
    p3 = x * x * x
    return p2, p3

x1 = 12
x2, x3 = square(x1)
print(x1, x2, x3)
```

```python
# 수치 출력 형식 지정
# f(foramt) 문자열 표시

a1 = 1.0 / 7.0
a2 = 123

str1 = f'a1={a1}, a2={a2}'
print(str1)
```

```python
# f 문자열의 상세 옵션

# .4f : 소수점 이하 네 자리 고정소수점 표시
# 04 : 정수를 0을 포함해 네 자리까지 표시

str2 = f'a1 = {a1:.4f} a2={a2:04}'
print(str2)
```

```python
str2 = f'a1 = {a1:.4f} a2={a2:04d}'
print(str2)
```

```python
# 04e : 소수점 이하 네 자리 부동소수점 표시
# #x : 정수를 16진수로 표시
str3 = f'a1 = {a1:.04e}  a2 = {a2:#x}'
print(str3)
```
