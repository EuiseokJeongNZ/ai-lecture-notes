## 2. Tensor란?

PyTorch에서 Tensor는 숫자를 담는 기본 컨테이너입니다.

NumPy의 `ndarray`와 비슷하지만, PyTorch Tensor는 딥러닝에 필요한 기능을 더 가지고 있습니다.

Tensor의 핵심 기능:

- 숫자 배열 저장
- 다차원 데이터 표현
- GPU 연산 지원
- 자동 미분 Autograd 지원

강의 핵심 표현:

```text
딥러닝 모델을 만든다는 것은 숫자로 이루어진 데이터를 변환하고 계산하는 과정이다.
```

````python
x = torch.tensor(1.0)

print(x)
print(type(x))
````

`torch.tensor(1.0)`은 숫자 하나를 Tensor로 만든 것입니다.

이것은 0차원 Tensor, 즉 스칼라 Tensor입니다.

## 3. 0차원 Tensor: Scalar

0차원 Tensor는 숫자 하나입니다.

예:

```text
1.0
5.0
```

딥러닝에서는 Loss 값처럼 숫자 하나로 표현되는 값을 스칼라 Tensor로 다룹니다.

````python
r0 = torch.tensor(1.0).float()

print(r0)
print()
print("type:", type(r0))
print("dtype:", r0.dtype)
print("shape:", r0.shape)
print("data:", r0.data)
````

코드 설명:

- `torch.tensor(1.0)`: 숫자 1.0을 Tensor로 생성합니다.
- `.float()`: 데이터 타입을 `float32`로 맞춥니다.
- `dtype`: Tensor 안 데이터 타입입니다.
- `shape`: Tensor의 모양입니다.

0차원 Tensor는 차원이 없기 때문에 `shape`이 `torch.Size([])`로 나옵니다.

## 4. 1차원 Tensor: Vector

1차원 Tensor는 숫자가 한 줄로 나열된 구조입니다.

예:

```text
[1, 2, 3, 4, 5]
```

수학에서는 벡터라고 부릅니다.

````python
r1_np = np.array([1, 2, 3, 4, 5])

print("NumPy 배열:")
print(r1_np)
print("NumPy shape:", r1_np.shape)

r1 = torch.tensor(r1_np).float()

print("\nTensor:")
print(r1)
print("dtype:", r1.dtype)
print("shape:", r1.shape)
````

코드 설명:

- `np.array(...)`: NumPy 배열을 만듭니다.
- `torch.tensor(r1_np)`: NumPy 배열을 PyTorch Tensor로 변환합니다.
- `.float()`: 실수형 Tensor로 바꿉니다.

`shape`이 `[5]`이면 원소가 5개인 1차원 Tensor입니다.

## 5. 2차원 Tensor: Matrix

2차원 Tensor는 행과 열이 있는 구조입니다.

예:

```text
[[1, 5, 6],
 [4, 3, 2]]
```

수학에서는 행렬 Matrix라고 부릅니다.

````python
r2_np = np.array([
    [1, 5, 6],
    [4, 3, 2]
])

print("NumPy 배열:")
print(r2_np)
print("NumPy shape:", r2_np.shape)

r2 = torch.tensor(r2_np).float()

print("\nTensor:")
print(r2)
print("shape:", r2.shape)
print("data:")
print(r2.data)
````

결과 해석:

```text
shape = torch.Size([2, 3])
```

- 2행
- 3열

즉 2x3 행렬입니다.

## 8. Tensor 데이터 타입 바꾸기

Tensor의 데이터 타입은 연산에 중요합니다.

대표 타입:

## 9. Tensor 형태 변환: `view()`

`view()`는 Tensor의 모양을 바꾸는 함수입니다.

NumPy의 `reshape()`와 비슷합니다.

단, 원소 총 개수는 유지되어야 합니다.

````python
print("원본 r3 shape:", r3.shape)
print("전체 원소 개수:", r3.numel())

r6 = r3.view(3, 4)

print("\nr3.view(3, 4):")
print(r6)
print("shape:", r6.shape)
````

결과 해석:

원래 `r3`의 shape는 `[3, 2, 2]`입니다.

전체 원소 개수:

```text
3 × 2 × 2 = 12
```

이를 `[3, 4]`로 바꾸면:

```text
3 × 4 = 12
```

원소 개수가 같기 때문에 가능합니다.

## 10. `view()`에서 `-1` 사용하기

`-1`은 해당 차원의 크기를 PyTorch가 자동으로 계산하라는 뜻입니다.

````python
r6_auto = r3.view(3, -1)

print(r6_auto)
print("shape:", r6_auto.shape)
````

`r3.view(3, -1)`에서 앞 차원은 3으로 고정합니다.

전체 원소가 12개이므로:

```text
3 × ? = 12
? = 4
```

그래서 shape가 `[3, 4]`가 됩니다.

## 11. 1차원으로 펼치기

`view(-1)`을 사용하면 Tensor를 1차원으로 펼칠 수 있습니다.

````python
r7 = r3.view(-1)

print(r7)
print("shape:", r7.shape)
````

`view(-1)`은 전체 원소를 한 줄로 펼칩니다.

원래 `r3`의 전체 원소 수가 12개이므로 결과는 `[12]`입니다.

## 12. NumPy `reshape()`와 PyTorch `view()` 비교

NumPy의 `reshape()`와 PyTorch의 `view()`는 비슷한 역할을 합니다.

````python
a_np = np.arange(6)

print("NumPy 원본:")
print(a_np)

print("\nNumPy reshape(2, 3):")
print(a_np.reshape(2, 3))

a_torch = torch.arange(6)

print("\nPyTorch 원본:")
print(a_torch)

print("\nPyTorch view(2, 3):")
print(a_torch.view(2, 3))
````

핵심:

- NumPy: `reshape()`
- PyTorch: `view()`

둘 다 원소 개수는 유지하면서 모양만 바꿉니다.

## 13. 전치 Tensor와 `view()` 주의점

Tensor를 전치하면 메모리 배치가 연속적이지 않을 수 있습니다.

이때 `view()`가 바로 안 될 수 있습니다.

````python
x = torch.randn(2, 3)
y = x.t()

print("x:")
print(x)
print("x shape:", x.shape)

print("\ny = x.t():")
print(y)
print("y shape:", y.shape)
print("y is contiguous?:", y.is_contiguous())
````

`.t()`는 2차원 Tensor의 행과 열을 바꿉니다.

하지만 전치된 Tensor는 메모리상 연속적이지 않을 수 있습니다.

그래서 이런 경우에는 `view()`보다 `reshape()`를 쓰는 것이 안전할 수 있습니다.

````python
try:
    print(y.view(6))
except RuntimeError as e:
    print("view 에러 발생:")
    print(e)

print("\nreshape 사용:")
print(y.reshape(6))
````

정리:

- `view()`는 메모리 배치가 연속적이어야 합니다.
- `reshape()`는 필요한 경우 내부적으로 복사해서 처리할 수 있습니다.

## 14. Tensor 주요 속성: `requires_grad`, `device`

Tensor에는 딥러닝 학습에 중요한 속성이 있습니다.

- `requires_grad`: 자동 미분 추적 여부
- `device`: Tensor가 CPU에 있는지 GPU에 있는지

````python
print("r1:", r1)
print("requires_grad:", r1.requires_grad)
print("device:", r1.device)
````

`requires_grad=False`이면 PyTorch가 이 Tensor의 연산을 미분 대상으로 추적하지 않습니다.

학습할 파라미터나 미분할 변수에는 `requires_grad=True`를 설정해야 합니다.

`item()`은 Tensor에서 Python 숫자를 꺼내는 함수입니다.

주로 Loss 값을 기록할 때 사용합니다.

````python
print("r0:", r0)

item_val = r0.item()

print("item type:", type(item_val))
print("item value:", item_val)
````

`r0`는 0차원 Tensor입니다.

`item()`을 사용하면 Tensor가 아니라 Python의 `float` 값으로 꺼낼 수 있습니다.

## 16. `item()` 사용 주의점

`item()`은 원소가 하나인 Tensor에만 사용할 수 있습니다.

````python
print("r1:")
print(r1)

try:
    print(r1.item())
except RuntimeError as e:
    print("item 에러 발생:")
    print(e)

t1 = torch.ones(1)

print("\n원소가 하나뿐인 1차원 Tensor:")
print(t1)
print("shape:", t1.shape)
print("item:", t1.item())
````

Loss처럼 숫자 하나짜리 Tensor를 Python 숫자로 바꿀 때 사용합니다.

## 17. Tensor의 최댓값 찾기

`max()`는 Tensor 안에서 가장 큰 값을 찾습니다.

다중 분류에서는 가장 큰 점수를 가진 클래스 인덱스를 찾을 때 자주 사용합니다.

````python
print("r2:")
print(r2)

print("\n전체 최댓값:")
print(r2.max())

print("\ntorch.max(r2):")
print(torch.max(r2))
````

인자 없이 `max()`를 호출하면 Tensor 전체에서 가장 큰 값을 반환합니다.

## 18. 축별 최댓값과 인덱스

`torch.max(r2, dim)` 형태로 쓰면 특정 축을 기준으로 최댓값을 구합니다.

반환값은 두 개입니다.

1. values: 최댓값
2. indices: 최댓값의 위치

````python
max_dim0 = torch.max(r2, 0)
max_dim1 = torch.max(r2, 1)

print("r2:")
print(r2)

print("\ndim=0 기준:")
print(max_dim0)

print("\ndim=1 기준:")
print(max_dim1)

print("\ndim=1 values:")
print(torch.max(r2, 1)[0])

print("\ndim=1 indices:")
print(torch.max(r2, 1)[1])
````

결과 해석:

- `dim=0`: 열 방향으로 비교합니다.
- `dim=1`: 행마다 최댓값을 찾습니다.

다중 분류 모델에서는 보통 각 행이 하나의 샘플이고, 각 열이 클래스 점수입니다.

그래서 `torch.max(output, 1)[1]`은 예측 클래스 번호를 구할 때 자주 사용됩니다.

## 20. 자동 미분 Autograd 개념

딥러닝에서 학습은 Loss를 줄이는 방향으로 파라미터를 수정하는 과정입니다.

이를 위해서는 미분값, 즉 경사 Gradient가 필요합니다.

PyTorch는 Autograd를 통해 이 미분을 자동으로 계산합니다.

자동 미분 흐름:

```text
requires_grad=True
→ Tensor 연산 수행
→ 계산 그래프 생성
→ backward()
→ .grad에서 경사 확인
→ zero_()로 초기화
```

````python
x_np = np.arange(-2, 2.1, 0.25)

print(x_np)
print("개수:", len(x_np))
````

`x_np`는 -2부터 2까지 0.25 간격으로 만든 NumPy 배열입니다.

이 값을 Tensor로 바꿔서 2차 함수의 기울기를 계산합니다.

## 21. `requires_grad=True` 설정

자동 미분을 하려면 Tensor를 만들 때 `requires_grad=True`를 설정해야 합니다.

````python
x = torch.tensor(
    x_np,
    requires_grad=True,
    dtype=torch.float32
)

print(x)
print("requires_grad:", x.requires_grad)
````

`requires_grad=True`는 PyTorch에게 다음 의미를 전달합니다.

```text
이 Tensor와 관련된 연산을 추적해줘.
나중에 미분값을 계산할 거야.
```

## 22. 2차 함수 계산하기

이번에는 다음 함수를 Tensor 연산으로 계산합니다.

```text
y = 2x² + 2
```

````python
y = 2 * x**2 + 2

print(y)
print("grad_fn:", y.grad_fn)
````

`grad_fn`은 이 Tensor가 어떤 연산으로 만들어졌는지를 나타냅니다.

`requires_grad=True`인 Tensor로 계산했기 때문에 PyTorch가 내부적으로 계산 그래프를 만들고 있습니다.

## 23. 2차 함수 그래프 그리기

`x.data`와 `y.data`를 사용해서 그래프를 그립니다.

````python
plt.plot(x.data, y.data)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = 2x^2 + 2")
plt.show()
````

그래프 해석:

`y = 2x² + 2`는 아래로 볼록한 U자 형태의 2차 함수입니다.

x가 0일 때 y가 가장 작고, x가 양쪽으로 멀어질수록 y가 커집니다.

## 24. `backward()`를 위해 스칼라로 만들기

PyTorch에서 `backward()`를 호출하려면 결과값이 스칼라여야 합니다.

현재 `y`는 여러 값을 가진 벡터이므로 `sum()`으로 하나의 값으로 만듭니다.

````python
z = y.sum()

print(z)
print("shape:", z.shape)
print("grad_fn:", z.grad_fn)
````

`z = y.sum()`은 y의 모든 값을 더해서 하나의 스칼라 Tensor를 만듭니다.

이제 `z.backward()`를 호출할 수 있습니다.

## 25. `backward()`로 경사 계산하기

`backward()`를 실행하면 PyTorch가 계산 그래프를 거꾸로 따라가며 미분합니다.

````python
z.backward()

print(x.grad)
````

결과 해석:

함수는 다음과 같습니다.

```text
y = 2x² + 2
```

미분하면:

```text
y' = 4x
```

따라서 `x.grad`는 각 x값에 4를 곱한 값과 같습니다.

## 26. 원래 함수와 경사 그래프 비교

원래 함수 y와 경사 x.grad를 함께 그려봅니다.

````python
plt.plot(x.data, y.data, label="y = 2x^2 + 2")
plt.plot(x.data, x.grad.data, label="gradient = 4x")
plt.legend()
plt.xlabel("x")
plt.ylabel("value")
plt.title("Function and Gradient")
plt.show()
````

그래프 해석:

- 파란 선은 원래 함수입니다.
- 검은 선은 기울기, 즉 gradient입니다.
- x가 음수일 때 gradient는 음수입니다.
- x가 양수일 때 gradient는 양수입니다.
- x가 0일 때 gradient는 0입니다.

즉, x=0 근처가 함수의 최저점이라는 것을 알 수 있습니다.

## 27. Gradient 누적 문제

PyTorch는 `backward()`를 호출할 때마다 gradient를 기존 값에 누적합니다.

그래서 새로 계산하기 전에는 gradient를 초기화해야 합니다.

````python
# 초기화하지 않고 한 번 더 backward를 실행해보기

y = 2 * x**2 + 2
z = y.sum()
z.backward()

print(x.grad)
````

결과를 보면 gradient가 이전보다 커져 있습니다.

이유는 PyTorch가 gradient를 덮어쓰지 않고 **누적**하기 때문입니다.

## 28. `zero_()`로 Gradient 초기화하기

새로운 미분 계산을 하기 전에는 `.grad.zero_()`로 gradient를 0으로 초기화해야 합니다.

````python
x.grad.zero_()

print(x.grad)
````

`zero_()` 뒤에 밑줄 `_`이 붙어 있습니다.

PyTorch에서 `_`가 붙은 함수는 보통 **원본 Tensor를 직접 수정하는 함수**입니다.

즉, `x.grad` 자체가 0으로 바뀝니다.

## 29. Sigmoid 함수 경사 계산: 내장 함수 사용

PyTorch에는 시그모이드 함수가 내장되어 있습니다.

```python
torch.nn.Sigmoid()
```

시그모이드는 출력값을 0과 1 사이로 바꾸는 함수입니다.

````python
sigmoid_layer = torch.nn.Sigmoid()

y_sigmoid = sigmoid_layer(x)

plt.plot(x.data, y_sigmoid.data)
plt.xlabel("x")
plt.ylabel("sigmoid(x)")
plt.title("Sigmoid Function")
plt.show()
````

그래프 해석:

시그모이드 함수는 S자 모양입니다.

- x가 작으면 0에 가까움
- x가 크면 1에 가까움
- x가 0 근처일 때 가장 빠르게 변함

## 30. Sigmoid의 gradient 계산

이번에도 `sum()`으로 스칼라를 만든 뒤 `backward()`를 호출합니다.

````python
z_sigmoid = y_sigmoid.sum()

z_sigmoid.backward()

print(x.grad)
````

`x.grad`에는 sigmoid 함수의 각 x 위치에서의 기울기가 저장됩니다.

시그모이드의 기울기는 가운데 근처에서 크고, 양 끝에서는 작아집니다.

## 31. Sigmoid 함수와 gradient 그래프 비교

시그모이드 함수와 그 gradient를 함께 그립니다.

````python
plt.plot(x.data, y_sigmoid.data, label="sigmoid(x)")
plt.plot(x.data, x.grad.data, label="gradient")
plt.legend()
plt.xlabel("x")
plt.ylabel("value")
plt.title("Sigmoid and Gradient")
plt.show()
````

그래프 해석:

- 시그모이드 함수는 S자 곡선입니다.
- gradient는 x=0 근처에서 가장 큽니다.
- 양 끝으로 갈수록 gradient가 작아집니다.

이 개념은 딥러닝에서 **기울기 소실 문제**와도 연결됩니다.

## 32. Sigmoid gradient 초기화

다음 계산을 위해 gradient를 다시 초기화합니다.

````python
x.grad.zero_()

print(x.grad)
````

`x.grad`가 모두 0이 되었습니다.

이후 다른 함수의 gradient를 새로 계산할 수 있습니다.

## 33. 직접 Sigmoid 함수 구현하기

이번에는 PyTorch 내장 함수가 아니라 직접 sigmoid 함수를 정의합니다.

공식:

```text
sigmoid(x) = 1 / (1 + exp(-x))
```

````python
def sigmoid(x):
    return 1 / (1 + torch.exp(-x))

y_custom_sigmoid = sigmoid(x)

plt.plot(x.data, y_custom_sigmoid.data)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Custom Sigmoid Function")
plt.show()
````

코드 설명:

- `torch.exp(-x)`: e의 -x 제곱을 계산합니다.
- `1 / (1 + torch.exp(-x))`: sigmoid 공식입니다.

직접 만든 함수도 Tensor 연산으로 구성되어 있기 때문에 자동 미분이 가능합니다.

## 34. 직접 만든 Sigmoid의 gradient 계산

직접 만든 sigmoid 함수도 `backward()`로 미분할 수 있습니다.

````python
z_custom = y_custom_sigmoid.sum()

z_custom.backward()

print(x.grad)
````

PyTorch는 직접 만든 함수라도 Tensor 연산으로 구성되어 있으면 계산 그래프를 자동으로 만듭니다.

그래서 `backward()`로 gradient를 계산할 수 있습니다.

## 35. 직접 만든 Sigmoid와 gradient 그래프

마지막으로 직접 만든 sigmoid 함수와 gradient를 함께 그립니다.

````python
plt.plot(x.data, y_custom_sigmoid.data, label="custom sigmoid")
plt.plot(x.data, x.grad.data, label="gradient")
plt.legend()
plt.xlabel("x")
plt.ylabel("value")
plt.title("Custom Sigmoid and Gradient")
plt.show()
````

그래프 해석:

내장 Sigmoid와 직접 만든 Sigmoid는 같은 형태입니다.

즉, PyTorch는 다음 두 경우 모두 자동 미분할 수 있습니다.

1. PyTorch 내장 함수 사용
2. Tensor 연산으로 직접 함수 구현

- Sigmoid의 gradient는 가운데에서 크고 양 끝에서 작습니다.
