응용예시

```python
# 두 변수를 비교해서 일치하는 개수로 정확도 계산

# 2개의 넘파이 배열 준비

# 정답
yt = np.array([1,1,0,1,0,1,1,0,1,0])

# 예측
yp = np.array([1,1,0,1,0,1,0,1,1,1])
```

```python
# 정답(yt)과 예측(yp) 잘 맞나? (비교)

(yt == yp)
```

```python
matched = (yt == yp)
matched.sum()
```

```python
# 맞은 개수(정답 수 세는 방법)
correct = matched.sum()

# 전체 수
total = len(matched)

# 정확도 계산
accuracy = correct / total
print(accuracy)
```

```python
print(f'정답수: {correct}, 전체 개수: {total}, 정확도: {accuracy*100}%')
```

벡터 값을 1차 함수로 변환 [0,1] 범위로 제한

```python
# 모든 값을 [0,1] 범위에 포함하도록 변환하고 싶어요
n1
```

```python
n1_max = n1.max()
n1_min = n1.min()
print(n1_max, n1_min)
```

```python
# 변환
(n1 - n1_min) / (n1_max - n1_min)
```
