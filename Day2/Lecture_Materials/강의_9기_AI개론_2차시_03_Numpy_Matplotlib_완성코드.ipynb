```python
# 집계함수 / axis 개념
# 집계함수

matrix = np.array([[1,2,3], [4,5,6]])
print(matrix)
```

```python
print('전체 합:', np.sum(matrix))
```

```python
print('열 기준 합(axis=0):', np.sum(matrix, axis=0)) # 세로방향
print('행 기준 합(axis=1):', np.sum(matrix, axis=1)) # 가로방향
```

```python
print('열 기준 평균(axis=0):', np.mean(matrix, axis=0))
```

```python
# matplotlib plot 기본 사용법
# 기본 Plot 실습
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label="Sin 함수")
plt.title("기본 선 그래프 (Line Plot)")
plt.xlabel("X 축")
plt.ylabel("Y 축")
plt.legend()
plt.show()
```

```python
# Scatter Plot 실습
plt.scatter(x, y, label="Sin 데이터 포인트")
plt.title("Scatter Plot 예제")
plt.xlabel("X 축")
plt.ylabel("Y 축")
plt.legend()
plt.show()
```

```python
# 실습 과제: 두 개의 함수를 하나의 그래프에 시각화하기

x = np.linspace(-5, 5, 200)
y1 = x            # 1차 함수
y2 = x ** 2       # 2차 함수

plt.plot(x, y1, label="y = x")
plt.plot(x, y2, label="y = x^2")
plt.title("1차 함수와 2차 함수 비교")
plt.xlabel("X 축")
plt.ylabel("Y 축")
plt.legend()
plt.grid(True)
plt.show()
```
